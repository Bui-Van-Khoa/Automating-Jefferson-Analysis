```python
import torchaudio
from speechbrain.pretrained import SpectralMaskEnhancement
```

```python
# Load the pre-trained model
model = SpectralMaskEnhancement.from_hparams(
    source="speechbrain/metricgan-plus-voicebank"
)

# Enhance the audio file
enhanced_audio = model.enhance_file("Data.wav")

# Save the enhanced audio to a new file
#torchaudio.save("enhanced_file.wav", enhanced_audio.squeeze(0).cpu(), 16000)
```

```python
import numpy as np
from scipy.io.wavfile import write

# Convert the tensor to a NumPy array
enhanced_audio_np = enhanced_audio.cpu().numpy()

# Save the enhanced audio as a WAV file
write("data_speechbrain.wav", 16000, enhanced_audio_np)
```
